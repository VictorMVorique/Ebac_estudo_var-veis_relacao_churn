# 3) Identifique e trate as colunas que contém outliers.
Caso opte por mante-los ou altera-los justifique sua escolha.
A identificação de outliers foi realizada utilizando o método do Intervalo Interquartil (IQR), considerando as variáveis numéricas Tempo_como_Cliente, Pagamento_Mensal e Total_Pago. Foram calculados os limites inferior e superior para cada variável e verificou-se que nenhuma observação ultrapassou esses limites. Os resultados mostraram 0 outliers em todas as variáveis analisadas, indicando que não foi necessário realizar qualquer tratamento, remoção ou ajuste dos dados. Os boxplots também corroboram essa conclusão, não apresentando observações extremas fora dos limites esperados

```python
colunas = ['Tempo_como_Cliente', 'Pagamento_Mensal', 'Total_Pago']

for col in colunas:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < limite_inferior) |
        (df[col] > limite_superior)
    ]

    print(f'{col}: {len(outliers)} outliers')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15,5))

for i, col in enumerate(colunas):
    sns.boxplot(y=df[col], ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```
